```python
from datetime import datetime
from zoneinfo import ZoneInfo

import numpy as np
import pandas as pd
from pandas import DataFrame

from vnpy.trader.constant import (Direction, Exchange, Offset, OptionType,
                                  OrderType, Product, Status)
from vnpy.trader.object import (CancelRequest, OrderData, PositionData,
                                TickData, TradeData)
```

```python
order_info_cols: dict[str, str] = {
    "symbol": 'string',
    "vt_symbol": 'string',
    "datetime": 'datetime64[ns, Asia/Shanghai]',
    "canceltime": 'datetime64[ns, Asia/Shanghai]',
    "exchange": 'object',  # enum: Exchange
    "ordersysid": 'string',
    "status": 'object',  # enum: Status
    "direction": 'object',  # enum: Direction
    "offset": 'object',  # enum: Offset
    "price": 'float64',
    "type": 'object',  # enum: OrderType
    "volume": 'float64',
    "traded": 'float64',
    "memo": 'string'
}

# 使用 order_info_columns 定义 DataFrame 的列类型
order_info: DataFrame = DataFrame(columns=list(order_info_cols.keys())).astype(order_info_cols)
```

```python
order_info
```

```python
order_info.dtypes
```

```python
CHINA_TZ = ZoneInfo("Asia/Shanghai")

# 随机生成一些 OrderData, 放到 order_data_list 中
order_data_list = [OrderData(
        symbol=f"IF{i+1:04d}",
        exchange=Exchange.CFFEX,
        orderid=f"a_{i+1:03d}_123",
        ordersysid=f"{i+1:010d}",
        direction=Direction.LONG if i % 2 == 0 else Direction.SHORT,
        offset=Offset.OPEN if i % 2 == 0 else Offset.CLOSE,
        price=3000.0 + i * 10,
        volume=1.0 + i * 0.1,
        type=OrderType.LIMIT,
        status=Status.NOTTRADED if i % 3 != 0 else Status.ALLTRADED,
        datetime=datetime.now().replace(tzinfo=CHINA_TZ),
        gateway_name="CTP",
    ) for i in range(10)]
```

```python
order_data_list[0].__dict__
```

```python
df = pd.concat(
    objs=[
        order_info,
        DataFrame(data=order_data_list, columns=list(order_info_cols.keys()))
    ],
    ignore_index=True)
# df = df.astype(order_info_cols)
df
```

### GroupBy

```python
speeds = pd.DataFrame(
    [
        ("bird", "Falconiformes", 389.0),
        ("bird", "Psittaciformes", 24.0),
        ("mammal", "Carnivora", 80.2),
        ("mammal", "Primates", np.nan),
        ("mammal", "Carnivora", 58),
    ],
    index=["falcon", "parrot", "lion", "monkey", "leopard"],
    columns=("class", "order", "max_speed"),
)
```

```python
speeds
```

```python
speeds.dtypes
```

```python
grouped = speeds.groupby(by="class")
group: DataFrame
for name, g in grouped:  # 每个 group (DataFrame) 都是一个 copy
    print(name)
    group = g

group.loc['lion', 'max_speed'] = 999.0  # 修改 copy 不会影响原始 speeds
group
```

```python
speeds
```

```python
speeds.loc[speeds['order'] == 'Falconiformes', 'max_speed'].item()
```
