### CTP作为底层接口时的持仓数据

首先, CTP 不支持"订阅持仓数据, 然后接收来自柜台的持仓数据实时更新".

持仓数据必须由客户端主动调用 `ReqQryInvestorPosition` 然后等待 `OnRspQryInvestorPosition` 回调.

而 VNPY 这边的实现是设置一个默认为 1 秒执行 1 次的定时任务 (在 EventEngine 里可有看到).

这也就导致了通过 `OmsEngine#get_all_positions` 获得的持仓数据并不总是最新的.
