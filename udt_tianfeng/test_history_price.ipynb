```python
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo
from pandas import DataFrame, Timestamp

import akshare as ak
import pandas as pd

CHINA_TZ = ZoneInfo("Asia/Shanghai")
```

```python
tool_trade_date_hist_sina_df = ak.tool_trade_date_hist_sina()
print(tool_trade_date_hist_sina_df.dtypes)
tool_trade_date_hist_sina_df.tail(10)
```

```python
tool_trade_date_hist_sina_df['trade_date'] = pd.to_datetime(tool_trade_date_hist_sina_df['trade_date']).dt.tz_localize(CHINA_TZ)
tool_trade_date_hist_sina_df.dtypes
```

```python
tool_trade_date_hist_sina_df.tail(10)
```

```python
datetime_now: datetime = pd.to_datetime(datetime.now(tz=CHINA_TZ)).replace(hour=0, minute=0, second=0, microsecond=0) + timedelta(days=1)
# 筛选出小于当前时间的交易日
matches: DataFrame = tool_trade_date_hist_sina_df.loc[tool_trade_date_hist_sina_df['trade_date'] < datetime_now]
# 筛选出最近的两天 (索引越大离现在越近)
matches: DataFrame = matches.tail(2).reset_index()

matches
```

```python
by_trade_date: Timestamp = matches.at[0, 'trade_date']
y_trade_date: Timestamp = matches.at[1, 'trade_date']
```

```python
byd_data: DataFrame = ak.get_futures_daily(start_date=by_trade_date.strftime("%Y%m%d"), end_date=by_trade_date.strftime("%Y%m%d"), market='CZCE')
byd_data
```

```python
yd_data: DataFrame = ak.get_futures_daily(start_date=y_trade_date.strftime("%Y%m%d"), end_date=y_trade_date.strftime("%Y%m%d"), market='CZCE') # type: ignore
yd_data
```
