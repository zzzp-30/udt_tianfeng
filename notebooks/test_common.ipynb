```python
from datetime import datetime as Datetime, timedelta
from zoneinfo import ZoneInfo

import numpy as np
import pandas as pd
from pandas import DataFrame, DatetimeIndex

from vnpy.trader.constant import (Direction, Exchange, Offset, OptionType,
                                  OrderType, Product, Status)
from vnpy.trader.object import (CancelRequest, OrderData, PositionData,
                                TickData, TradeData)
```

```python
CHINA_TZ = ZoneInfo("Asia/Shanghai")
```

```python
end_date = Datetime(2025, 9, 30, tzinfo=CHINA_TZ)
```

```python
def calculate_remaining_trading_days(end_date: Datetime) -> int:
    """计算剩余交易日"""
    current_time = Datetime.now(tz=CHINA_TZ)
    return sum(
        1 for day in range((end_date - current_time).days + 1)
        if (current_time + timedelta(day)).weekday() < 5
    )

calculate_remaining_trading_days(end_date)
```

```python
def calculate_remaining_trading_days(end_date: Datetime) -> int:
    """计算剩余交易日"""
    current_time: Datetime = Datetime.now(tz=CHINA_TZ)
    business_days: DatetimeIndex = pd.bdate_range(
        start=current_time.date(),
        end=end_date.date(),
        freq='B',  # 'B' 表示工作日频率
        tz='Asia/Shanghai',
        inclusive='right'  # 不包含开始日期, 仅包含结束日期, 即 (current_date. end_date]
    )
    
    return len(business_days)

calculate_remaining_trading_days(end_date)
```
