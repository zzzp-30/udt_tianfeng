```python
from time import sleep
from vnpy.utility.cooldown import Cooldown, CooldownMap, StackableCooldownMap
```

```python
cooldown_map = CooldownMap(base=Cooldown(timeout_seconds=1.0))
stackable_cooldown_map = StackableCooldownMap(base=Cooldown(timeout_seconds=10.0), stacks=3)

for i in range(10):
    print(f"test_{i}", cooldown_map.test("key") and stackable_cooldown_map.test("key"))
    sleep(1.0)
    
```
